### Derived Features

We'll have to derive some features from our separate dataframes and then combine those together.

Let's create a dataframe of stat descriptions for each region:

```python
stream_region_describe_df = stream_temps_all_df.groupby(['region'])['max_celcius'].describe()
```

```python
salmon_region_describe_df = salmon_all_df.groupby(['region'])['stream_observed_count'].describe()
```

```python
pressures_region_lo_describe_df = pressures_all_df.groupby(['region'])['cumulative_lopct'].describe()
```

```python
pressures_region_mod_describe_df = pressures_all_df.groupby(['region'])['cumulative_modpct'].describe()
```

```python
pressures_region_hi_describe_df = pressures_all_df.groupby(['region'])['cumulative_hipct'].describe()
```

### Merge

```python
dfs = {
    'max_celcius': stream_region_describe_df, 
    'stream_observed_count': salmon_region_describe_df, 
    'cumulative_lopct': pressures_region_lo_describe_df,
    'cumulative_modpct': pressures_region_mod_describe_df,
    'cumulative_hipct': pressures_region_hi_describe_df
}
```

```python
region_stats_df = pd.concat(dfs, axis=1)
```
